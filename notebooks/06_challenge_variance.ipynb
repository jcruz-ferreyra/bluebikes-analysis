## Methodology and Objectives

Standard software assumes your data meets certain asymptotic requirements. This challenge asks the you to prove if your specific dataset actually supports the math you are
using. Derivation of the Variance of the Sample Mean. Instead of just reporting $\bar{X}$. You are asked to derive the theoretical variance of the sample mean for your chosen process (e.g., if you chose an AR(1), you must show $\text{Var}(\bar{X}) = \frac{\sigma^2}{n} \cdot \frac{1 + \phi}{1 - \phi}$).).

You should then perform a Bootstrap or Sub-sampling analysis: take the dataset, break it into 20 overlapping windows, calculate the mean of each, and see if the variance of those 20 means matches the theoretical formula you derived. If the empirical variance is much higher than the theory predicts, you must write a critical analysis on why your model (and standard p-values) might be overconfident due to high persistence or fat tails.

### Theoretical Variance Derivation: SARIMA(1,1,1)(1,0,0)₇

Following model selection analysis ([notebook](04_classical_model_selection.ipynb)), we identified SARIMA(1,1,1)(1,0,0)₇ as the optimal specification for both pickup and dropoff morning demand. After first differencing ($d=1$), the process follows an ARMA(1,1) structure with seasonal AR(1) at weekly lag:

$$\nabla Y_t = \phi_1 \nabla Y_{t-1} + \Phi_1 \nabla Y_{t-7} + \varepsilon_t + \theta_1 \varepsilon_{t-1}$$

where:
- $\phi_1$ is the non-seasonal AR(1) parameter (captures day-to-day persistence)
- $\theta_1$ is the MA(1) parameter (captures error correction from yesterday)
- $\Phi_1$ is the seasonal AR(1) parameter at lag 7 (captures weekly patterns)
- $\varepsilon_t \sim N(0, \sigma^2)$ is white noise

For the differenced ARMA(1,1) component (ignoring seasonal complexity for derivation), the theoretical variance of the sample mean is:

$$\text{Var}(\bar{X}) = \frac{\sigma^2}{n} \cdot \frac{(1+\theta_1)^2 + \phi_1^2 - 2\phi_1\theta_1}{(1-\phi_1)^2}$$

The correction factor $\frac{(1+\theta_1)^2 + \phi_1^2 - 2\phi_1\theta_1}{(1-\phi_1)^2}$ adjusts for autocorrelation structure, increasing variance when positive autocorrelation reduces effective sample size or decreasing it when MA and AR parameters offset. We extract fitted parameters ($\hat{\phi}_1$, $\hat{\theta}_1$, $\hat{\sigma}^2$) from the trained model and compute the theoretical variance, then compare against empirical variance from overlapping subsample means to test whether asymptotic theory holds for our finite sample with potential model misspecification.
