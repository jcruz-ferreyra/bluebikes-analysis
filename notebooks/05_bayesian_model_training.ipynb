## Bayesian Time Series Model: Uncertainty-Aware Forecasting

We extend the forecasting comparison by introducing a Bayesian time series model for Bluebikes demand. Unlike SARIMA, which estimates fixed model parameters, and Prophet, which provides a structured additive decomposition, the Bayesian approach treats the main model components as uncertain quantities. Trend, weekly seasonality, and residual variation are represented through probability distributions and updated using the observed demand data.

The model follows a regression-style time series formulation, where demand is explained through an intercept, a trend component, and seasonal features constructed from Fourier terms. The Fourier features are fixed inputs derived from time, while the model estimates the posterior distribution of their weights. This allows the model not only to produce a central forecast, but also to quantify uncertainty around future demand.

This approach is useful for understanding how Bayesian inference works in forecasting. Instead of returning a single fitted trend or seasonal pattern, the model returns a range of plausible trends, seasonal effects, and future demand trajectories. We use the same train-test or cross-validation structure applied to the SARIMA and Prophet models so that predictive performance and uncertainty calibration can be compared consistently.

The inspiration for the bayesian exploration comes from [this](https://www.pymc-labs.com/blog-posts/probabilistic-forecasting-optimization-under-uncertainty) blog post.
